## **¿Por qué se dice que una MTU es capaz de simular cualquier Máquina de Turing?**

Una **Máquina de Turing Universal (MTU)** es capaz de simular cualquier Máquina de Turing porque recibe como entrada:

- La **descripción codificada** de una Máquina de Turing $M$, representada como $\langle M \rangle$.
- Una **cadena de entrada** $w$ sobre la cual debe ejecutarse dicha máquina.

A partir de esta información, la MTU interpreta la descripción de $M$ y reproduce paso a paso su funcionamiento sobre la cadena $w$.

Formalmente:

$$
U(\langle M,w\rangle)=M(w)
$$

Por lo tanto, la MTU no está construida para resolver un único problema específico, sino que puede ejecutar distintas Máquinas de Turing según la máquina que se le proporcione codificada como entrada.
Esta característica permite considerar a la MTU como una **máquina de propósito general**, ya que una misma máquina puede ejecutar diferentes algoritmos representados mediante distintas codificaciones.

## **Dada la siguiente MT M:** 

| Q   | 0        | 1        |
| --  | --       | --       |
| q0  | (q1,1,R) | (q1,0,R) |
| q1  | (qf,0,R) | (qf,1,R) |
| qf  | -        | -        |

#### ¿Qué hace $M$? 
 
La máquina comienza en el estado $q0$ y modifica el primer símbolo que lee: 
 
- Si lee `0`, escribe `1`. 
- Si lee `1`, escribe `0`. 
 
En ambos casos avanza una posición hacia la derecha y pasa al estado $q1$. 
 
En $q1$: 
 
- Si lee `0`, escribe `0`. 
- Si lee `1`, escribe `1`. 
 
El segundo símbolo leído se mantiene sin modificaciones. Después de procesarlo, la máquina avanza hacia la derecha y pasa al estado final $qf$. 
En resumen, la máquina $M$ **invierte el primer símbolo que procesa, mantiene sin cambios el siguiente símbolo y luego finaliza**. 
 

#### ¿Qué información debería recibir una MTU para poder simular $M$? 
 
Para poder simular a $M$, la Máquina de Turing Universal necesita recibir: 
 
- La descripción codificada de la máquina $M$, representada como $\langle M\rangle$. 
- La configuración de la cinta de $M$. 
- El estado actual de $M$. 
- El símbolo que está siendo leído por la cabeza. 
- La codificación de las transiciones de $M$. 
 
Para representar toda esta información es necesario codificar: 
 
- Los símbolos de la cinta. 
- Los estados. 
- Los movimientos de la cabeza. 
- Las transiciones de la máquina. 
 
La entrada general de la MTU puede representarse como: 
 
$$
\langle M,w\rangle
$$
 
donde $\langle M\rangle$ contiene la descripción codificada de la máquina y $w$ representa el contenido de su cinta. 

#### 3. Codificación de la cinta de la MTU 
 
La configuración dada para la cinta de $M$ es: 
 
`1 q0 0 1 1` 
 
Esto significa que: 
 
- El estado actual es $q0$. 
- La cabeza está posicionada sobre el símbolo `0`. 
- A la izquierda de la cabeza se encuentra `1`. 
- A la derecha se encuentra `11`. 
 
Para representar esta información en la cinta de la MTU se utiliza la siguiente codificación. 
 
##### Codificación de los estados 
 
- $q0 = 00$ 
- $q1 = 01$ 
- $qf = 10$ 
 
##### Codificación de los símbolos 
 
- $0 = 0$ 
- $1 = 1$ 
 
##### Codificación de los movimientos 
 
- $R = 0$ 
- $L = 1$ 
 
Las transiciones de $M$ son: 
 
$$
\delta(q0,0)=(q1,1,R)
$$ 
 
$$
\delta(q0,1)=(q1,0,R)
$$ 
 
$$
\delta(q1,0)=(qf,0,R)
$$ 
 
$$
\delta(q1,1)=(qf,1,R)
$$ 
 
La codificación de cada transición se obtiene concatenando el estado actual, el símbolo leído, el nuevo estado, el símbolo escrito y el movimiento de la cabeza.
 
| Transición | Codificación | 
|---|---| 
| $(q0,0,q1,1,R)$ | `0000110` | 
| $(q0,1,q1,0,R)$ | `0010100` | 
| $(q1,0,qf,0,R)$ | `0101000` | 
| $(q1,1,qf,1,R)$ | `0111010` | 
 
En la configuración: 
 
`1 q0 0 1 1` 
 
la cabeza está leyendo el `0`. Para representar esta situación en la cinta de la MTU, el símbolo leído se ubica después del separador `$`, junto con la codificación del estado actual. 
 
Por lo tanto, la configuración queda representada como: 
 
`1*11$000` 
 
donde: 
 
- `*` indica la posición de la cabeza. 
- `00` representa el estado actual $q0$. 
- `0` representa el símbolo que está leyendo la cabeza. 
 
A continuación se agrega la codificación de las transiciones, utilizando `#` como separador entre los distintos registros. 
 
La cinta codificada de la MTU queda: 
 
`1*11$000#0000110#0010100#0101000#0111010` 
 
De esta manera, la parte izquierda de `$` representa la cinta y la posición de la cabeza de $M$, mientras que la parte derecha contiene el estado actual, el símbolo leído y la descripción codificada de las transiciones necesarias para que la MTU pueda simular a $M$.

## 1. **Codificación de una máquina simple**  
   * Definir una máquina $M$ que ...
   * Codificar sus estados, símbolos y transiciones en forma numérica

Recibe una cadena binaria formada por símbolos `0` y `1` y **reemplaza cada bit por su complemento**.

La máquina realiza dos etapas:

1. Recorre la cadena de izquierda a derecha invirtiendo cada bit:
   - Si encuentra `0`, escribe `1`.
   - Si encuentra `1`, escribe `0`.
2. Cuando encuentra el blanco `_` que marca el final de la cadena, comienza a desplazarse hacia la izquierda sin modificar los símbolos.

Por ejemplo, para la entrada:

`01001`

la cadena resultante será:

`10110`

---

### Estados

La máquina utiliza dos estados:

- $qStart$: recorre la cadena hacia la derecha e invierte cada bit.
- $qBack$: recorre la cadena hacia la izquierda sin modificarla.

Los nombres de los estados se utilizan para describir claramente su función. Para la codificación numérica se utilizan sus correspondientes códigos binarios.

---

### Tabla de transiciones

| Q      | 0            | 1            | _           |
|--------|--------------|--------------|-------------|
| qStart | (qStart,1,R) | (qStart,0,R) | (qBack,_,L) |
| qBack  | (qBack,0,L)  | (qBack,1,L)  | -           |

Las transiciones son:

$$
\delta(qStart,0)=(qStart,1,R)
$$

$$
\delta(qStart,1)=(qStart,0,R)
$$

$$
\delta(qStart,\_)=(qBack,\_,L)
$$

$$
\delta(qBack,0)=(qBack,0,L)
$$

$$
\delta(qBack,1)=(qBack,1,L)
$$

Cuando la máquina se encuentra en $qBack$ y lee `_`, no existe una transición definida. Por lo tanto, la ejecución se detiene.

---

### Codificación numérica

#### Estados

- $qStart = 00$
- $qBack = 01$

El estado inicial es:

`00`

#### Símbolos

- `0 = 00`
- `1 = 01`
- `_ = 10`

#### Movimientos

- `R = 0`
- `L = 1`

---

### Codificación de las transiciones

Cada transición se representa concatenando:

`estado actual + símbolo leído + nuevo estado + símbolo escrito + movimiento`

Cada estado ocupa 2 bits, cada símbolo ocupa 2 bits y el movimiento ocupa 1 bit.

Por lo tanto, cada transición tiene:

$$
2+2+2+2+1=9\ bits
$$

| Transición | Codificación |
|---|---|
| $(qStart,0,qStart,1,R)$ | `000000010` |
| $(qStart,1,qStart,0,R)$ | `000100000` |
| $(qStart,\_,qBack,\_,L)$ | `001001101` |
| $(qBack,0,qBack,0,L)$ | `010001001` |
| $(qBack,1,qBack,1,L)$ | `010101011` |

## 2. **Simulación básica**  
   * Implementar en Python un programa que reciba:
     * La codificación de una máquina $M$
     * Una cadena de entrada $w$
   * El programa debe simular paso a paso la ejecución de $M$ sobre $w$

### Codificación completa de la máquina

Para que el simulador pueda determinar cuál es el estado inicial sin conocer previamente los nombres de los estados, la codificación comienza indicando dicho estado.

Se utiliza `$` para separar el estado inicial de la lista de transiciones:

`estado inicial $ transiciones`

Las distintas transiciones se separan mediante `#`.

La codificación completa de $M$ es:

`00$000000010#000100000#001001101#010001001#010101011`

donde:

- `00` antes de `$` indica el estado inicial.
- `$` separa el estado inicial de la descripción de las transiciones.
- `#` separa las distintas transiciones.

El simulador trabaja directamente con estos códigos, por lo que no necesita conocer los nombres de los estados.

````python
# ============================================================
# CÓDIGOS UTILIZADOS POR LA MTU
# ============================================================

COD_SIMBOLOS = {
    "00": "0",
    "01": "1",
    "10": "_"
}

COD_MOVIMIENTOS = {
    "0": "R",
    "1": "L"
}


# ============================================================
# VALIDACIÓN DE LA MÁQUINA CODIFICADA
# ============================================================

def validar_codificacion(codificacion):

    # La codificación no puede estar vacía
    if not codificacion:
        return False, (
            "Error: la codificación de la máquina "
            "no puede estar vacía."
        )

    # Solo se permiten los símbolos utilizados
    for caracter in codificacion:
        if caracter not in "01#$":
            return False, (
                f"Error: se encontró el carácter '{caracter}'. "
                "La codificación solo puede contener "
                "0, 1, # y $."
            )

    # Debe existir exactamente un separador $
    if codificacion.count("$") != 1:
        return False, (
            "Error: la codificación debe contener "
            "exactamente un símbolo $."
        )

    # Separamos estado inicial y transiciones
    estado_inicial, parte_transiciones = codificacion.split("$")

    # El estado inicial debe tener 2 bits
    if len(estado_inicial) != 2:
        return False, (
            "Error: el estado inicial debe estar "
            "codificado utilizando 2 bits."
        )

    # Una máquina puede no tener transiciones
    if parte_transiciones == "":
        return True, "Máquina correctamente codificada."

    registros = parte_transiciones.split("#")

    # No puede haber transiciones vacías
    for i, registro in enumerate(registros, start=1):
        if registro == "":
            return False, (
                f"Error: la transición {i} está vacía. "
                "Revise la posición de los separadores #."
            )

    # Validamos cada transición
    for i, registro in enumerate(registros, start=1):

        # 2 estado + 2 símbolo + 2 estado +
        # 2 símbolo + 1 movimiento = 9 bits
        if len(registro) != 9:
            return False, (
                f"Error: la transición {i} tiene "
                f"{len(registro)} bits. "
                "Cada transición debe contener "
                "exactamente 9 bits."
            )

        simbolo_leido_cod = registro[2:4]
        simbolo_escrito_cod = registro[6:8]
        movimiento_cod = registro[8]

        if simbolo_leido_cod not in COD_SIMBOLOS:
            return False, (
                f"Error en la transición {i}: "
                f"el símbolo leído '{simbolo_leido_cod}' "
                "no está definido."
            )

        if simbolo_escrito_cod not in COD_SIMBOLOS:
            return False, (
                f"Error en la transición {i}: "
                f"el símbolo escrito '{simbolo_escrito_cod}' "
                "no está definido."
            )

        if movimiento_cod not in COD_MOVIMIENTOS:
            return False, (
                f"Error en la transición {i}: "
                f"el movimiento '{movimiento_cod}' "
                "no está definido."
            )

    return True, "Máquina correctamente codificada."


# ============================================================
# VALIDACIÓN DE LA PALABRA
# ============================================================

def validar_entrada(entrada):

    if not entrada:
        return False, (
            "Error: la palabra de entrada no puede estar vacía."
        )

    for caracter in entrada:
        if caracter not in "01":
            return False, (
                f"Error: la palabra contiene el símbolo "
                f"'{caracter}'. "
                "La entrada debe estar formada únicamente "
                "por 0 y 1."
            )

    return True, "Palabra válida."


# ============================================================
# DECODIFICACIÓN DE LA MÁQUINA
# ============================================================

def decodificar_maquina(codificacion):

    transiciones = {}

    # Obtenemos el estado inicial de la propia codificación
    estado_inicial, parte_transiciones = codificacion.split("$")

    # Si la máquina no tiene transiciones,
    # devolvemos únicamente su estado inicial
    if parte_transiciones == "":
        return estado_inicial, transiciones

    registros = parte_transiciones.split("#")

    for registro in registros:

        estado_actual = registro[0:2]
        simbolo_leido_cod = registro[2:4]
        nuevo_estado = registro[4:6]
        simbolo_escrito_cod = registro[6:8]
        movimiento_cod = registro[8]

        simbolo_leido = COD_SIMBOLOS[simbolo_leido_cod]
        simbolo_escrito = COD_SIMBOLOS[simbolo_escrito_cod]
        movimiento = COD_MOVIMIENTOS[movimiento_cod]

        transiciones[(estado_actual, simbolo_leido)] = (
            nuevo_estado,
            simbolo_escrito,
            movimiento
        )

    return estado_inicial, transiciones


# ============================================================
# MOSTRAR CONFIGURACIÓN
# ============================================================

def mostrar_configuracion(paso, cinta, cabeza, estado):

    representacion = cinta.copy()

    # El * indica la posición del cabezal
    representacion.insert(cabeza, "*")

    print(
        f"Paso {paso}: "
        f"{''.join(representacion)}   "
        f"Estado: {estado}"
    )


# ============================================================
# SIMULACIÓN
# ============================================================

def simular(estado_inicial, transiciones, entrada):

    # Cinta inicial
    cinta = ["_"] + list(entrada) + ["_"]

    # El cabezal comienza sobre el primer símbolo de w
    cabeza = 1

    # El estado inicial se obtiene de la máquina codificada
    estado = estado_inicial

    paso = 0

    mostrar_configuracion(
        paso,
        cinta,
        cabeza,
        estado
    )

    while True:

        simbolo = cinta[cabeza]

        clave = (estado, simbolo)

        # Si no existe una transición aplicable,
        # la máquina simulada se detiene
        if clave not in transiciones:
            print(
                "\nNo existe una transición definida para "
                f"(estado {estado}, símbolo {simbolo})."
            )
            print("La máquina se detiene.")
            break

        nuevo_estado, simbolo_escrito, movimiento = \
            transiciones[clave]

        # Escribimos el nuevo símbolo
        cinta[cabeza] = simbolo_escrito

        # Movimiento del cabezal
        if movimiento == "R":

            cabeza += 1

            if cabeza == len(cinta):
                cinta.append("_")

        elif movimiento == "L":

            if cabeza == 0:
                cinta.insert(0, "_")
                cabeza = 0
            else:
                cabeza -= 1

        # Actualizamos el estado
        estado = nuevo_estado

        paso += 1

        mostrar_configuracion(
            paso,
            cinta,
            cabeza,
            estado
        )

    resultado = "".join(cinta).strip("_")

    print(f"\nEntrada:   {entrada}")
    print(f"Resultado: {resultado}")


# ============================================================
# PROGRAMA PRINCIPAL
# ============================================================

while True:

    maquina_codificada = input(
        "Ingrese la máquina codificada: "
    ).strip()

    es_valida, mensaje = validar_codificacion(
        maquina_codificada
    )

    print(mensaje)

    if es_valida:
        break

    print("Ingrese nuevamente la máquina.\n")


while True:

    entrada = input(
        "Ingrese la palabra de entrada: "
    ).strip()

    es_valida, mensaje = validar_entrada(
        entrada
    )

    print(mensaje)

    if es_valida:
        break

    print("Ingrese nuevamente la palabra.\n")


# Decodificamos la máquina
estado_inicial, transiciones = decodificar_maquina(
    maquina_codificada
)


# Ejecutamos la simulación
print("\n--- Inicio de la simulación ---\n")

simular(
    estado_inicial,
    transiciones,
    entrada
)
````

### Reflexión: relación entre la MTU y las computadoras modernas

La Máquina de Turing Universal introduce una idea fundamental de la computación: **una misma máquina puede ejecutar distintos procedimientos si recibe una descripción codificada de esos procedimientos**.

Una Máquina de Turing común tiene sus transiciones determinadas de antemano y realiza una tarea específica. En cambio, una MTU recibe como información tanto la máquina que debe simular como la entrada sobre la que debe ejecutarla.

En nuestro simulador se observa esta separación claramente:

* La **máquina codificada** contiene las instrucciones.
* La **palabra de entrada** contiene los datos que serán procesados.
* El **simulador** interpreta las instrucciones y las ejecuta paso a paso.

Esta idea se relaciona directamente con el funcionamiento de las computadoras modernas. Una computadora física no necesita ser construida de manera diferente para ejecutar cada programa. El mismo hardware puede ejecutar un editor de texto, un navegador, un videojuego o un compilador porque cada programa se encuentra representado mediante información que la computadora puede interpretar y ejecutar.

De manera similar, nuestro simulador no contiene directamente la lógica de la máquina que invierte los bits. Esa lógica se encuentra en:

`00$000000010#000100000#001001101#010001001#010101011`

Si se proporciona otra máquina correctamente codificada utilizando la misma convención, el simulador puede interpretar otras transiciones sin modificar el mecanismo general de ejecución.

La MTU permite así separar dos conceptos:

**Máquina que ejecuta** y **Programa que describe qué debe ejecutar**

Esta separación constituye una de las ideas centrales de las computadoras de propósito general: **las instrucciones pueden representarse como datos y ser interpretadas por una misma máquina**.

Por eso, aunque una Máquina de Turing Universal sea un modelo teórico muy simple comparado con una computadora actual, ambas comparten el mismo principio fundamental: una única máquina puede realizar diferentes cálculos dependiendo del programa que recibe.
