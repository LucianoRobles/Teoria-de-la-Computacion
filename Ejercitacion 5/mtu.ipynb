## **¿Por qué se dice que una MTU es capaz de simular cualquier Máquina de Turing?**

Una **Máquina de Turing Universal (MTU)** es capaz de simular cualquier Máquina de Turing porque recibe como entrada:

- La **descripción codificada** de una Máquina de Turing $M$, representada como $\langle M \rangle$.
- Una **cadena de entrada** $w$ sobre la cual debe ejecutarse dicha máquina.

A partir de esta información, la MTU interpreta la descripción de $M$ y reproduce paso a paso su funcionamiento sobre la cadena $w$.

Formalmente:

$$
U(\langle M,w\rangle)=M(w)
$$

Por lo tanto, la MTU no está construida para resolver un único problema específico, sino que puede ejecutar distintas Máquinas de Turing según la máquina que se le proporcione codificada como entrada.
Esta característica permite considerar a la MTU como una **máquina de propósito general**, ya que una misma máquina puede ejecutar diferentes algoritmos representados mediante distintas codificaciones.

Sea $M$ una Máquina de Turing que acepta todas las cadenas que terminan en `01`.

La Máquina de Turing Universal $U$ recibe la descripción de $M$, representada por $\langle M\rangle$, junto con una cadena de entrada $w$, y simula el comportamiento de $M$ sobre dicha cadena.

#### $U(\langle M\rangle,1101)$

La cadena `1101` termina en `01`.

$$
U(\langle M\rangle,1101)=acepta
$$

#### $U(\langle M\rangle,100)$

La cadena `100` no termina en `01`.

$$
U(\langle M\rangle,100)=rechaza
$$

#### $U(\langle M\rangle,01)$

La cadena `01` termina en `01`.

$$
U(\langle M\rangle,01)=acepta
$$

#### $U(\langle M\rangle,111)$

La cadena `111` no termina en `01`.

$$
U(\langle M\rangle,111)=rechaza
$$`

La MTU obtiene estos resultados porque simula el comportamiento de $M$ para cada una de las cadenas de entrada, cumpliendo asi su función.

| Entrada | ¿Termina en `01`? | Resultado |
|---|---|---|
| `1101` | Sí | Acepta |
| `100` | No | Rechaza |
| `01` | Sí | Acepta |
| `111` | No | Rechaza |

##**Dada la siguiente MT M:** 

| Q   | 0        | 1        |
| --  | --       | --       |
| q0  | (q1,1,R) | (q1,0,R) |
| q1  | (qf,0,R) | (qf,1,R) |
| qf  | -        | -        |

1. Explicar que hace M 
2. Explicar qué información debería recibir una MTU para poder simular M
3. Codificar la cintar de MTU sabiendo que configuración de la cinta de MT M es 1 q0 0 1 1

#### ¿Qué hace $M$? 
 
La máquina comienza en el estado $q0$ y modifica el primer símbolo que lee: 
 
- Si lee `0`, escribe `1`. 
- Si lee `1`, escribe `0`. 
 
En ambos casos avanza una posición hacia la derecha y pasa al estado $q1$. 
 
En $q1$: 
 
- Si lee `0`, escribe `0`. 
- Si lee `1`, escribe `1`. 
 
El segundo símbolo leído se mantiene sin modificaciones. Después de procesarlo, la máquina avanza hacia la derecha y pasa al estado final $qf$. 
En resumen, la máquina $M$ **invierte el primer símbolo que procesa, mantiene sin cambios el siguiente símbolo y luego finaliza**. 
 

#### ¿Qué información debería recibir una MTU para poder simular $M$? 
 
Para poder simular a $M$, la Máquina de Turing Universal necesita recibir: 
 
- La descripción codificada de la máquina $M$, representada como $\langle M\rangle$. 
- La configuración de la cinta de $M$. 
- El estado actual de $M$. 
- El símbolo que está siendo leído por la cabeza. 
- La codificación de las transiciones de $M$. 
 
Para representar toda esta información es necesario codificar: 
 
- Los símbolos de la cinta. 
- Los estados. 
- Los movimientos de la cabeza. 
- Las transiciones de la máquina. 
 
La entrada general de la MTU puede representarse como: 
 
$$
\langle M,w\rangle
$$
 
donde $\langle M\rangle$ contiene la descripción codificada de la máquina y $w$ representa el contenido de su cinta. 

#### 3. Codificación de la cinta de la MTU 
 
La configuración dada para la cinta de $M$ es: 
 
`1 q0 0 1 1` 
 
Esto significa que: 
 
- El estado actual es $q0$. 
- La cabeza está posicionada sobre el símbolo `0`. 
- A la izquierda de la cabeza se encuentra `1`. 
- A la derecha se encuentra `11`. 
 
Para representar esta información en la cinta de la MTU se utiliza la siguiente codificación. 
 
##### Codificación de los estados 
 
- $q0 = 00$ 
- $q1 = 01$ 
- $qf = 10$ 
 
##### Codificación de los símbolos 
 
- `0 = 0` 
- `1 = 1` 
 
##### Codificación de los movimientos 
 
- `R = 0` 
- `L = 1` 
 
Las transiciones de $M$ son: 
 
$$
\delta(q0,0)=(q1,1,R)
$$ 
 
$$
\delta(q0,1)=(q1,0,R)
$$ 
 
$$
\delta(q1,0)=(qf,0,R)
$$ 
 
$$
\delta(q1,1)=(qf,1,R)
$$ 
 
La codificación de cada transición se obtiene concatenando el estado actual, el símbolo leído, el nuevo estado, el símbolo escrito y el movimiento de la cabeza.
 
| Transición | Codificación | 
|---|---| 
| $(q0,0,q1,1,R)$ | `0000110` | 
| $(q0,1,q1,0,R)$ | `0010100` | 
| $(q1,0,qf,0,R)$ | `0101000` | 
| $(q1,1,qf,1,R)$ | `0111010` | 
 
En la configuración: 
 
`1 q0 0 1 1` 
 
la cabeza está leyendo el `0`. Para representar esta situación en la cinta de la MTU, el símbolo leído se ubica después del separador `$`, junto con la codificación del estado actual. 
 
Por lo tanto, la configuración queda representada como: 
 
`1*11$000` 
 
donde: 
 
- `*` indica la posición de la cabeza. 
- `00` representa el estado actual $q0$. 
- `0` representa el símbolo que está leyendo la cabeza. 
 
A continuación se agrega la codificación de las transiciones, utilizando `#` como separador entre los distintos registros. 
 
La cinta codificada de la MTU queda: 
 
`1*11$000#0000110#0010100#0101000#0111010` 
 
De esta manera, la parte izquierda de `$` representa la cinta y la posición de la cabeza de $M$, mientras que la parte derecha contiene el estado actual, el símbolo leído y la descripción codificada de las transiciones necesarias para que la MTU pueda simular a $M$.
